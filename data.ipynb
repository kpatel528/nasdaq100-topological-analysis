```python
import pandas as pd

df = pd.read_csv("nasdaq100_latest_raw_data.csv")
df.head()
```

```python
df[df.isna().all(axis=1)]
```

```python
df['date'] = pd.to_datetime(df['date'], utc=True).dt.tz_localize(None).dt.normalize()
df = df.sort_values(['date', 'Company'])
df = df.drop(columns=['Date'])
df = df.rename(columns={'date': 'Date'})
df.head()
```

```python
prices = df.pivot(index='Date', columns='Company', values='Adj_Close')
min_years = 5
min_days = min_years * 252
prices = prices.dropna(axis=1, thresh=min_days)
df['Company'].unique()
```

```python
df = df.sort_values(['Company', 'Date'])
df['Return'] = df.groupby('Company')['Adj_Close'].pct_change()
df = df.dropna(subset=['Return'])
df.head()
```

```python
df.to_csv('cleaned_stock_data.csv', index=False)
```
